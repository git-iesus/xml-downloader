```python
# Importações
import webbrowser
import pyautogui as ptg
import pandas as pd
import keyboard

# Configurações
ptg.PAUSE = 1
ptg.FAILSAFE = True
df = pd.read_excel('dados.xlsx')

# Funções
def validar_img(imagem):
    posicao_imagem = None
    while posicao_imagem is None:
        try:
            posicao_imagem = ptg.locateOnScreen(imagem, confidence=0.7)

        except ptg.ImageNotFoundException:
            pass

        ptg.sleep(1)
```
